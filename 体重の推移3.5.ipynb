```python
# 必要なライブラリを読み込み
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path as path
from sklearn.linear_model import LinearRegression
from sklearn import set_config

set_config(display="text")
```

```python
# CSVを読み込み
input_path =path("体重計記録_前処理済み.csv")

df = pd.read_csv(input_path)
```

```python
# 体重と体脂肪率の相関関係を計算
correlation = df["体重 (kg)"].corr(df["体脂肪率 %"])

print(f"相関係数：{correlation:.3f}")


# 説明変数と目的変数を用意
X = df[["体重 (kg)"]]
y = df["体脂肪率 %"]

print()
print("Xの形：", X.shape)
print("yの形：", y.shape)


# 線形回帰モデルを作成して学習
model = LinearRegression()
model.fit(X, y)
```

```python
# 日本語の文字化けを防ぐ
plt.rcParams["font.family"] = "meiryo"


# 体重と体脂肪率の散布図
fig, ax = plt.subplots(figsize=(8,6))

ax.scatter(
    df["体重 (kg)"],   #横軸
    df["体脂肪率 %"],  #縦軸
    s=18,             #点のサイズ
    alpha=0.4
)


# 回帰直線用の予測値を作成
X_line = X.sort_values("体重 (kg)")
y_pred = model.predict(X_line)


# 回帰直線
ax.plot(
    X_line["体重 (kg)"],
    y_pred,
    color="red",
    linewidth=2,
    label="回帰直線",
    alpha=0.7
)


ax.set_title(
    f"体重と体脂肪率の関係性 (相関係数  r = {correlation:.3f})"
)

ax.set_xlabel("体重 (kg)")
ax.set_ylabel("体脂肪率 (%)")

ax.grid(alpha=0.3)
ax.legend()

plt.tight_layout()
plt.show()
```

```python
# 体重と筋肉率の相関を計算
m_correlation = df["体重 (kg)"].corr(df["筋肉率 %"])

print(f"相関係数：{m_correlation:.3f}")
```

```python
# 体重と筋肉量の相関を計算
mm_correlation = df["体重 (kg)"].corr(df["筋肉量 (kg)"])

print(f"相関係数：{mm_correlation:.3f}")
```

```python
# 目的変数を用意
y_rate = df["筋肉率 %"]
y_mass = df["筋肉量 (kg)"]

# それぞれの回帰モデルを作成して学習
rate_model = LinearRegression()
mass_model = LinearRegression()

rate_model.fit(X, y_rate)
mass_model.fit(X, y_mass)
```

```python
# 回帰直線用の予測値を作成
X_line = X.sort_values("体重 (kg)")

rate_pred = rate_model.predict(X_line)
mass_pred = mass_model.predict(X_line)


# 散布図を左右に作成
fig, (ax_rate, ax_mass) = plt.subplots(
    1, 2,
    figsize=(14, 6)
)


# 左：体重と筋肉率
ax_rate.scatter(
    df["体重 (kg)"],
    y_rate,
    s=18,
    alpha=0.5
)

ax_rate.plot(
    X_line["体重 (kg)"],
    rate_pred,
    color="red",
    linewidth=2,
    label="回帰直線"
)

ax_rate.set_title(f"体重と筋肉率 (r = {m_correlation:.3f}) ")
ax_rate.set_xlabel("体重 (kg)")
ax_rate.set_ylabel("筋肉率 (%)")
ax_rate.grid(alpha=0.3)
ax_rate.legend()


# 右：体重と筋肉量
ax_mass.scatter(
    df["体重 (kg)"],
    y_mass,
    s=18,
    alpha=0.5
)

ax_mass.plot(
    X_line["体重 (kg)"],
    mass_pred,
    color="red",
    linewidth=2,
    label="回帰直線"
)

ax_mass.set_title(f"体重と筋肉量 (r = {mm_correlation:.3f})")
ax_mass.set_xlabel("体重 (kg)")
ax_mass.set_ylabel("筋肉量 (kg)")
ax_mass.grid(alpha=0.3)
ax_mass.legend()


plt.tight_layout
plt.show()
```
